### Price Prediction 

```python
import pandas as pd
import joblib
import os

# Load the test data
test_data_path = "/Users/ARahim/Documents/GitHub/immo-eliza-ml/Data/test_house.csv"
test_data = pd.read_csv(test_data_path)

# Load the trained model
model_directory = "/Users/ARahim/Documents/GitHub/immo-eliza-ml"
model_path = os.path.join(model_directory, 'artifacts.joblib')
loaded_model = joblib.load(model_path)

# Prepare the test data for prediction
X_test = test_data.drop(columns=['price'])

# Make predictions
predictions = loaded_model.predict(X_test)

# Add the predictions to the test data
test_data['predicted_price'] = predictions

# Print the predictions or save them to a CSV file
print(test_data[['price', 'predicted_price']])

# Save the predictions to a CSV file
predictions_output_path = "/Users/ARahim/Documents/GitHub/immo-eliza-ml/Data/predictions.csv"
test_data[['price', 'predicted_price']].to_csv(predictions_output_path, index=False)
```
