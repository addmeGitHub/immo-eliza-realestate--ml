###  Model (4) Random Forest Model

```python
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
import joblib
import os

# Load the CSV file into a DataFrame
file_path = "/Users/ARahim/Documents/GitHub/immo-eliza-ml/Data/properties.csv"
properties_df = pd.read_csv(file_path)

# Split the DataFrame based on the 'property_type' column
df_house = properties_df[properties_df['property_type'] == 'HOUSE']

# Split df_house into training and testing sets
train_house, test_house = train_test_split(df_house, test_size=0.2, random_state=42)

# Define the numerical and categorical features
num_features = ["total_area_sqm", "surface_land_sqm", "nbr_frontages", 
                "nbr_bedrooms", "terrace_sqm", "garden_sqm", 
                "primary_energy_consumption_sqm"]

fl_features = ["fl_furnished", "fl_open_fire", "fl_terrace", "fl_garden", "fl_swimming_pool", "fl_floodzone",
               "fl_double_glazing"]

cat_features = ["subproperty_type", "region", "province", "locality", 
                "equipped_kitchen", "state_building", "epc", "heating_type"]

# Imputation of missing values for categorical features in the training set
cat_imputer = SimpleImputer(strategy="most_frequent")
train_house[cat_features] = cat_imputer.fit_transform(train_house[cat_features])

# Imputation of missing values for categorical features in the test set
test_house[cat_features] = cat_imputer.transform(test_house[cat_features])

# Imputation of missing values for numerical features in the training set
num_imputer = SimpleImputer(strategy="median")
num_imputer.fit(train_house[num_features])
train_house[num_features] = num_imputer.transform(train_house[num_features])

# Imputation of missing values for numerical features in the test set
test_house[num_features] = num_imputer.transform(test_house[num_features])

# One-hot encoding for categorical features
enc = OneHotEncoder(handle_unknown='ignore')
enc.fit(train_house[cat_features])
train_house_cat = enc.transform(train_house[cat_features]).toarray()
test_house_cat = enc.transform(test_house[cat_features]).toarray()

# Combine the numerical and one-hot encoded categorical columns for training set
X_train = pd.concat([
    train_house[num_features + fl_features].reset_index(drop=True),
    pd.DataFrame(train_house_cat, columns=enc.get_feature_names_out())
], axis=1)
y_train = train_house['price']

# Combine the numerical and one-hot encoded categorical columns for testing set
X_test = pd.concat([
    test_house[num_features + fl_features].reset_index(drop=True),
    pd.DataFrame(test_house_cat, columns=enc.get_feature_names_out())
], axis=1)
y_test = test_house['price']

print("Features: \n", X_train.columns.tolist())

# Save the transformed dataframes for later use
train_file_path = "/Users/ARahim/Documents/GitHub/immo-eliza-ml/Data/train_house.csv"
test_file_path = "/Users/ARahim/Documents/GitHub/immo-eliza-ml/Data/test_house.csv"

# Concatenate features and target variable for train and test sets
train_df = pd.concat([X_train.reset_index(drop=True), y_train.reset_index(drop=True)], axis=1)
test_df = pd.concat([X_test.reset_index(drop=True), y_test.reset_index(drop=True)], axis=1)

# Save train and test dataframes to CSV
train_df.to_csv(train_file_path, index=False, header=True)  # Ensure header=True to preserve column names
test_df.to_csv(test_file_path, index=False, header=True)  # Ensure header=True to preserve column names

# Load the CSV files into DataFrames
df_train = pd.read_csv(train_file_path)
df_test = pd.read_csv(test_file_path)

# Identify and drop outliers for numeric features using IQR
def remove_outliers(df, columns, tolerance=1.5):
    for column in columns:
        q1 = df[column].quantile(0.25)
        q3 = df[column].quantile(0.75)
        iqr = q3 - q1
        lower_bound = q1 - tolerance * iqr
        upper_bound = q3 + tolerance * iqr
        df = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    return df

# Remove outliers from the training data
df_train = remove_outliers(df_train, num_features)

# Remove outliers from the testing data using the same tolerance as the training data
df_test = remove_outliers(df_test, num_features)

# Prepare the data for modeling
X_train = df_train.drop(columns=['price'])
y_train = df_train['price']
X_test = df_test.drop(columns=['price'])
y_test = df_test['price']

# Train the Random Forest regression model with cross-validation
random_forest_model = RandomForestRegressor(n_estimators=100, random_state=42)  # Use 100 decision trees
cv_scores = cross_val_score(random_forest_model, X_train, y_train, cv=5)  # 5-fold cross-validation
mean_cv_score = cv_scores.mean()

# Print mean cross-validation score
print("Mean Cross-Validation Score:", mean_cv_score)

# Train the Random Forest regression model
random_forest_model.fit(X_train, y_train)

# Evaluate the model
train_score = random_forest_model.score(X_train, y_train)
test_score = random_forest_model.score(X_test, y_test)

print("Train Score:", train_score)
print("Test Score:", test_score)

# Save the directory path
directory = "/Users/ARahim/Documents/GitHub/immo-eliza-ml"

# Create the directory if it doesn't exist
if not os.path.exists(directory):
    os.makedirs(directory)

# Save the model artifacts in the specified directory
joblib.dump(random_forest_model, os.path.join(directory, 'artifacts.joblib'))

# Now let's try to load the saved model
loaded_model = joblib.load(os.path.join(directory, 'artifacts.joblib'))
```
